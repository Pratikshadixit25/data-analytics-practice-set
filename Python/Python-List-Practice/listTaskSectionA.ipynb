```python
# 1. Print all elements from the list.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
print(numbers)
#using for loop
for num in numbers:
    print(num)
```

```python
#2. Print the first and last element.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
print(numbers[0:len(numbers):9])#using slicing
#below using indexing
print(numbers[0])#first element
print(numbers[-1])#last element
```

```python
#3. Find the total number of elements without manually counting them.
#using Logic
count=0
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
for num in numbers:
    count+=1
print(count)
#using len function
len(numbers)
```

```python

#4. Calculate the sum of all elements.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
sum_iterable=0
for num in numbers:
    sum_iterable+=num
print(sum_iterable)

```

```python
#5. Find the largest and smallest number.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#using min() and max() method
print(min(numbers))
print(max(numbers))
#logical method:
largest=numbers[0]
smallest=numbers[0]
for num in numbers:
    if num>largest:
        largest=num
    elif num<smallest:
        smallest=num
print("largest number",largest)
print("smallest number",smallest)
    

```

```python
#print all even numbers
numbers=[12,45,7,23,89,12,34,5,67]
even_numbers=[]
for num in numbers:
    if num%2==0:
        even_numbers.append(num)
print("List of Even Numbers: ",even_numbers)
```

```python
#print all odd numbers
numbers=[12,45,7,23,89,12,34,5,67]
odd_numbers=[]
for num in numbers:
    if num%2==1:
        odd_numbers.append(num)
print("List of Odd numbers",odd_numbers)
```

```python
#Count how many even and odd numbers are present
numbers=[12,45,7,23,89,12,34,5,67]
count_even=0
count_odd=0

for num in numbers:
    if num%2==0:
        count_even+=1
    else:
        count_odd+=1
print(count_even)
print(count_odd)
```

```python
#check whether 23 exists in the list
numbers=[12,45,7,23,89,12,34,5,67]
#using in opeartor
if 23 in numbers:
    print(f"Yes Exists..")
else:
    print("Not Exists..")
    
#using for loop
for num in numbers:
    if num==23:
        print(num,"Exists in List")
        
```

```python
#Find the index position of 89
numbers=[12,45,7,23,89,12,34,5,67]
#using index method
print(numbers.index(89))

#logical
count=0
for num in numbers:
    if num==89:
        break
    count+=1
print(count)
    
```

```python
#: Add 100 at the end of the list
numbers=[12,45,7,23,89,12,34,5,67]
numbers.append(100)
print(numbers)
```

```python
#Insert 50 at index position 3.
numbers=[12, 45, 7, 23, 89, 12, 34, 5, 67, 100]

numbers.insert(3,50)
print(numbers)
```

```python
#Remove the first occurrence of 45
numbers=[12, 45, 7, 50, 23, 89, 12, 34, 5, 67, 100]
numbers.remove(45)
print(numbers)
```

```python
#Remove duplicate values from the list.
#using set
numbers=[12, 7, 50, 23, 89, 12, 34, 5, 67, 100]

# unique_numbers=set(numbers)
# print(unique_numbers) --> it returns unique values in set
    
#using dict
dict={}
for num in numbers:
    dict[num]=num
print(dict) #it returns unique values in dictioanry
    
#using dict method  
data=list(dict.fromkeys(numbers))
print(data)#it returns unique values in list but opeartion performed using dictionary method
```

```python
#Sort the list in ascending and descending order
numbers=[12, 7, 50, 23, 89, 12, 34, 5, 67, 100]
#using sort method for ascnding 
numbers.sort()
print(numbers)
#sorted method for reverse
print(sorted(numbers,reverse=True))
```
