```python
#Merge two lists without using the + operator.

numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
#using extend method
numbers.extend(students)
print(numbers)

numbers1= [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
students1 = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
#uisng logic
for elements in students1:
    numbers1.append(elements)
print(numbers1)
```

```python
#Find common elements between two lists.
list1 = [4, 12, "apple", 7, 19, "banana", 33]
list2 = ["cherry", 7, 45, 12, "date", 99, 4]

#method 1
for data in list1:
    if data in list2:
        print(data)
#method 2 using set
print(set(list1) & set(list2))

```

```python
#Find elements present in the first list but not in the second.
list1 = [4, 12, "apple", 7, 19, "banana", 33]
list2 = ["cherry", 7, 45, 12, "date", 99, 4]

#using logic
for data in list1:
    if data not in list2:
        print(data)

#using set
result=set(list1)-set(list2)
print(result)
```

```python
#Rotate a list one position to the right.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#method 1
numbers[-1]
numbers.insert(0,numbers[-1])
numbers.pop(-1)
print(numbers)

numbers1 = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#method 2
rotate=[numbers1[-1]] + numbers1[:-1]
print(rotate)
```

```python
#Rotate a list by a user-given n positions
numbers=[67, 12, 45, 7, 23, 45, 89, 12, 34, 5]

n=int(input("Enter a number"))
rotation = numbers[-n:] + numbers[:-n]
print(rotation)
```

```python
#Move all zero values to the end of a list
numbers = [0, 12, 0, 45, 7, 0, 23, 89, 0, 34]

zeros=[]
non_zeros=[]
for num in numbers:
    if num==0:
        zeros.append(num)
    else:
        non_zeros.append(num)

result=non_zeros + zeros
print(result)
```

```python
#Divide a list into smaller lists of three elements each
numbers = [12, 45, 7, 23, 89, 12, 34, 5, 67, 100, 3]

small_list=[]

for num in range(0,len(numbers),3):
    small_list.append(numbers[num:num+3])
print(small_list) 
```

```python
#Check whether a list is a palindrome.
my_list = [5, 12, 34, 56, 78, 78, 56, 34, 12, 5]

#using logic
if my_list==my_list[::-1]:
    print("List is palindrome")
else:
    print("list is not Palindrome")
#using one line
result=my_list==my_list[::-1]
print(result)

#using for loop
for num in my_list:
    if num==my_list[::-1]:
        print("yes list is palindrome")

```

```python
#Find two numbers whose sum equals a user-given target.

numbers = [2, 7, 11, 15, 3, 6]
target = int(input("Enter A number"))

for num in range(len(numbers)):
    for j in range(num+1,len(numbers)):
        if numbers[num]+numbers[j]==target:
            print(numbers[num],numbers[j])
        
```

#above explanation: Outer loop, num = 0 (pointing at value 2)

Inner loop: range(0+1, 6) → range(1, 6) → j takes values 1, 2, 3, 4, 5
So we check: numbers[0]+numbers[1] (2+7), numbers[0]+numbers[2] (2+11), numbers[0]+numbers[3] (2+15), numbers[0]+numbers[4] (2+3), numbers[0]+numbers[5] (2+6)
Notice: j never becomes 0 — so we never compare numbers[0] with itself.

Outer loop, num = 1 (pointing at value 7)

Inner loop: range(1+1, 6) → range(2, 6) → j takes values 2, 3, 4, 5
Checks: numbers[1]+numbers[2] (7+11), numbers[1]+numbers[3] (7+15), numbers[1]+numbers[4] (7+3), numbers[1]+numbers[5] (7+6)
Notice: j never goes back to 0 (which would recheck 7+2, the same pair as 2+7 from before, just flipped) — and j never equals 1 either (comparing numbers[1] to itself).
