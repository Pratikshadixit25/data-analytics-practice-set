```python
#Print all student names.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
print(students)
for name in students:
    print(name)
```

```python
#Print names that start with "A".
students = ["Rahul", "Priya", "amit", "Sneha", "Raj", "Anjali"]

for name in students:
    if name.lower().startswith("a"):
        print(name)
        
#using logical 
for names in students:
    if names[0].lower()=="a":
        print(names)
```

```python
#Print names containing more than five characters.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

for name in students:
    if len(name)>5:
        print(name)
#logical method:
for name in students:
    count=0
    for char in name:
        count+=1
        if count>5:
            print(name)
            break
```

```python
# Convert every name to uppercase.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
for name in students:
    up=name.upper()
    print(up)
```

```python
#Sort the names alphabetically.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
students.sort()
print(students)
```

```python
#Find the longest student name
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

longest_name=students[0]

for name in students:
    if len(name)>len(longest_name):
        longest_name=name
print(longest_name)
    
```

```python
#Check whether "Sneha" exists in the list.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
#using in opeartor
if "Sneha" in students:
    print("Yes")
    
#using logic
check=False
for name in students:
    if name=="Sneha":
        check=True
        break
if check:
    print("yes")
else:
    print("no")
```

```python
#Create a new list containing the length of every name.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

li=[]
for name in students:
    size=len(name)
    li.append(size)
print(li)
```

```python
#Remove "Raj" from the list.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

#using logic
for name in students:
    if name!="Raj":
        print(name)
        
#using method:
students.remove("Raj")
print(students)
```

```python
#Print all names in reverse alphabetical order.
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]

reversed_names=sorted(students,reverse=True)
print(reversed_names)
    
```
