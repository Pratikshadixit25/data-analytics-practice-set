```python
#Create a new list containing the square of every number.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#using logic
sq=[]
for num in numbers:
    square_number=num**2
    sq.append(square_number)
print(sq)
#comprehension method
sq=[num**2 for num in numbers]
print(sq)

```

```python
#Create a new list containing only numbers greater than 30
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#using logic
result=[]
for num in numbers:
    if num>30:
        result.append(num)
print(result)
#using list comherension

result=[num for num in numbers if num>30]
print(result)
```

```python
#Print the list in reverse order without using reverse()
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#by using slicing
print(numbers[::-1])
#using loop
for i in range(len(numbers)-1,-1,-1):
    print(numbers[i])
```

```python
#Find the second-largest number
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]

#first we will remove duplicates
unique_values=set(numbers)
print(unique_values) #we get unique values
#now sort data in desceding order to get second largest
descending_order=sorted(unique_values,reverse=True)
print(descending_order[1])

#one line code
result=sorted(set(numbers),reverse=True)
print(result[1])
```

```python
#Find the second-smallest number
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
numbers.sort()#here i know already small numbers are not repeted thats why i used sort 
print(numbers[1])
#in case if we have duplicates like 5 repeted that time we have to remove duplicates and then sort in ascdeing order

unique=set(numbers)
final_output=sorted(unique)
print(final_output[1])
```

```python
#Count how many times each number appears.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]

#importing counter form collection module
from collections import Counter
print(Counter(numbers))

#logical method : below we wrote counter code 
repeted_numbers={}
for num in numbers:
    if num in repeted_numbers:
        repeted_numbers[num]+=1 #here we update the value of dictioanry when number is added again 
    else:
        repeted_numbers[num]=1 #here repeted numbers does not come so assign one 
print(repeted_numbers)
```

```python
#Separate the numbers into two lists — even numbers and odd numbers.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
even_li=[]
odd_li=[]

for num in numbers:
    if num%2==0:
        even_li.append(num)
    else:
        odd_li.append(num)
print("even numbers",even_li)
print("odd numbers",odd_li)
```

```python
#Find all duplicate elements.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
#create counter code for practice
repete_num={}
for num in numbers:
    if num in repete_num:
        repete_num[num]+=1
    else:
        repete_num[num]=1


li=[]
for k,v in repete_num.items():
    if v>1:
        li.append(k)
print("dupliate values",li)
```

```python
#Find all elements that appear only once.
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]

key_value_pairs={}
for num in numbers:
    if num in key_value_pairs:
        key_value_pairs[num]+=1
    else:
        key_value_pairs[num]=1

li=[]       
for k,v in key_value_pairs.items():
    if v==1:
        li.append(k)
print(li)      
    
```

```python
#Replace every number smaller than 20 with 0
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]


for num in range(len(numbers)):
    if numbers[num]<20:
        numbers[num]=0 
print(numbers)
        
        
```
